# Pilot report

Reads the run ledger and summaries written by `scripts/00`–`05`. **It contains no results**: outputs are produced only when you run it.

* Records with `synthetic: true` (experiment IDs `SYN-*`) are integration tests and must not be reported as findings.
* Failed and skipped runs are listed, not hidden.
* Thresholds in `prereg/decision_gates.md` stay UNRESOLVED until this pilot is reviewed.

```python
import json
from collections import Counter
from pathlib import Path

RUNS = Path('..') / 'runs'
ledger = [json.loads(l) for l in (RUNS / 'ledger.jsonl').read_text(encoding='utf-8').splitlines() if l]
print(len(ledger), 'runs;', Counter((r['status'], r['synthetic']) for r in ledger))
```

```python
# Failed / skipped runs (must be discussed in the pilot write-up)
for r in ledger:
    if r['status'] != 'ok':
        rec = json.loads((RUNS / r['path']).read_text(encoding='utf-8'))
        print(r['experiment_id'], r['status'], rec.get('skipped_steps') or rec.get('error', '')[-200:])
```

```python
# Gate 0 / intrinsic / ablation tables as printed by the scripts
for p in sorted(RUNS.glob('*-summary*/*.txt')):
    print('=' * 20, p.parent.name, p.name)
    print(p.read_text(encoding='utf-8'))
```

```python
# Proposed follow-up hypotheses (status 'proposed'; literature unverified)
fu = RUNS / 'followups.jsonl'
if fu.exists():
    for line in fu.read_text(encoding='utf-8').splitlines():
        h = json.loads(line)
        print(h['hypothesis_id'], '| synthetic' if h['synthetic'] else '', '\n  ', h['observed_evidence'])
```
